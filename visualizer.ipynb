```python
import folium
from folium.plugins import MarkerCluster
from branca.colormap import LinearColormap
import random

def generate_random_color():
    return f"#{random.randint(0, 0xFFFFFF):06x}"

frame = zod_data['001538']
print(zod_data['001538'].latitude,zod_data['001538'].longitude)
sequences = frame.getSequences()
location = [frame.latitude, frame.longitude]  # Latitude, Longitude
m = folium.Map(location=location, zoom_start=20, tiles='CartoDB positron')  # Start zoom (adjust as needed)

drawBoundingBox(m, location)

# Create a color map for sequences
sequence_colors = {seq_id: generate_random_color() for seq_id in sequences.keys()}

# Create a legend
legend_html = f'''
    <div style="position: fixed; bottom: 50px; left: 50px; width: 220px; height: auto; background-color: white; 
                border: 2px solid grey; z-index:9999; font-size:14px; padding: 10px">
    <b>Frame ID: {frame.frame_id}</b><br>
    Sequences:<br>
'''

for seq_id, sequence in sequences.items():
    color = sequence_colors[seq_id]
    if sequence is not None:
        num_images = len(sequence.images)
        legend_html += f'<span style="color:{color};">●</span> {seq_id}: {num_images} images<br>'
        addMapillarySequenceToMap(m, sequence, color=color)
    else:
        legend_html += f'<span style="color:{color};">●</span> {seq_id}: No images available<br>'


legend_html += '</div>'

# Add the legend to the map
m.get_root().html.add_child(folium.Element(legend_html))

# Marker (Keep this for testing)
folium.Marker(location=location, popup="Test Point").add_to(m)

# Display Map
m
# m.save("001538.html")
```
